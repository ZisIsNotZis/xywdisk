```python
import numpy as np
X = np.memmap('real4.47f,34f', '47f,34f', 'r', shape=9999)
Y = X['f1']
X = X['f0']
I = np.array(Y.any(1).argmin())
```

```python
print('='*80)
print('disk', (X[I-1,:7]*100).astype('H'))
print('='*80)
print('ptm', (X[I-1,7:]*400).astype('H'))
print('='*80)
```

```python
X[np.linalg.norm(Y[:I],axis=1).argmin()]-X[I-1]
```

```python
import matplotlib.pyplot as plt
from sklearn.linear_model import Ridge
w = np.linalg.lstsq(X[:I]-X[:I].mean(0),Y[:I]-Y[:I].mean(0))[0]
logw = np.log(abs(w))
logw = np.sign(w)*(logw-logw.max(1,keepdims=1)+5).clip(0)
X[I-1,w[7:,Y[I-1].argsort()[1]].argsort()[-3:]]
```

```python
# w = Ridge(.5).fit(X[:I], Y[:I]).coef_.T
plt.figure(figsize=(16,4))
plt.imshow(logw[:7], 'rainbow')
plt.colorbar()
```
